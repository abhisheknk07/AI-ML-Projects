#                            Plant Disease Identifier

#1. Introduction

Plant diseases can affect crop health, quality, and yield. Early identification of visible symptoms on leaves can help farmers take suitable action before a disease spreads. Manual identification requires observation and agricultural knowledge. Image-based machine learning can provide a simple automated approach by learning visual differences between healthy and diseased leaves.

This project develops a Plant Disease Identifier using image features and the K-Nearest Neighbors (KNN) algorithm. A subset of the PlantVillage dataset is used. Four tomato leaf classes are selected: healthy, Early Blight, Late Blight, and Septoria Leaf Spot. The images are resized and converted into numerical color and texture-related features before training the KNN classifier.

#2. Problem Statement

The main objective of this project is to classify a plant leaf image into the correct health or disease category using image processing and machine learning.

The input is a leaf image. Image features such as RGB statistics, grayscale statistics, and color histograms are extracted from the image. These numerical features are given to a KNN classifier to predict the corresponding class.


#3. Results and Discussion

The project evaluates the KNN classifier using accuracy, precision, recall, and F1 score. A confusion matrix is used to identify which classes are classified correctly and which classes are confused with one another.

The project also generates a class distribution graph and sample prediction images. These visualizations help explain the dataset and show examples of correct and incorrect predictions.

```python
# 1. Install required libraries
# ------------------------------------------------------------

!pip -q install -U datasets

# ------------------------------------------------------------
# 2. Import libraries
# ------------------------------------------------------------

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from PIL import Image
from datasets import load_dataset

from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

print("Libraries imported successfully!")
print("TensorFlow version:", tf.__version__)


# ============================================================
# PROJECT INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("        PLANT DISEASE IDENTIFIER")
print("        USING KNN AND CNN")
print("=" * 70)

# Four classes selected from PlantVillage
CLASS_NAMES = [
    "Tomato___healthy",
    "Tomato___Early_blight",
    "Tomato___Late_blight",
    "Tomato___Septoria_leaf_spot"
]

print("\nSelected classes:")

for i, name in enumerate(CLASS_NAMES):
    print(f"{i}: {name}")


# ============================================================
# 3. LOAD THE PARQUET-BASED PLANTVILLAGE DATASET
# ============================================================


print("\nLoading PlantVillage dataset...")
print("First run may take some time.")

dataset = load_dataset(
    "geraldmc/plantvillage-full",
    revision="v0.1.0"
)

print("\nDataset loaded successfully!")

print(dataset)


# ============================================================
# 4. CREATE TRAIN AND TEST DATASETS
# ============================================================

full_ds = dataset["train"]

print("\nTotal rows:", len(full_ds))


# ------------------------------------------------------------
# Filter four selected tomato classes
# ------------------------------------------------------------

selected_train = full_ds.filter(
    lambda row:
        row["class_label"] in CLASS_NAMES
        and row["split"] == "train"
)

selected_test = full_ds.filter(
    lambda row:
        row["class_label"] in CLASS_NAMES
        and row["split"] == "test"
)

print(
    "\nSelected training images:",
    len(selected_train)
)

print(
    "Selected testing images:",
    len(selected_test)
)


# ============================================================
# 5. LIMIT DATASET SIZE FOR GOOGLE COLAB
# ============================================================

MAX_TRAIN_PER_CLASS = 500
MAX_TEST_PER_CLASS = 150


def select_balanced_data(
    dataset,
    max_per_class
):

    selected_rows = []

    for class_name in CLASS_NAMES:

        class_data = dataset.filter(
            lambda row:
                row["class_label"] == class_name
        )

        # Shuffle deterministically
        class_data = class_data.shuffle(
            seed=42
        )

        number_to_take = min(
            max_per_class,
            len(class_data)
        )

        selected_rows.extend(
            [
                class_data[i]
                for i in range(number_to_take)
            ]
        )

    return selected_rows


train_rows = select_balanced_data(
    selected_train,
    MAX_TRAIN_PER_CLASS
)

test_rows = select_balanced_data(
    selected_test,
    MAX_TEST_PER_CLASS
)


print("\nFinal training images:", len(train_rows))
print("Final testing images :", len(test_rows))


# ============================================================
# 6. CREATE LABEL MAPPING
# ============================================================

label_to_id = {
    name: i
    for i, name in enumerate(CLASS_NAMES)
}

id_to_label = {
    i: name
    for name, i in label_to_id.items()
}


# ============================================================
# 7. IMAGE PREPROCESSING
# ============================================================

IMG_SIZE = (64, 64)


def preprocess_image(
    image
):

    # Convert to RGB
    image = image.convert("RGB")

    # Resize
    image = image.resize(
        IMG_SIZE
    )

    # Convert to NumPy
    image = np.array(
        image,
        dtype=np.float32
    )

    # Normalize to 0-1
    image = image / 255.0

    return image


# ============================================================
# 8. LOAD IMAGES INTO NUMPY ARRAYS
# ============================================================

print("\nProcessing training images...")

X_train_img = np.array([
    preprocess_image(
        row["image"]
    )
    for row in train_rows
])


print("Processing testing images...")

X_test_img = np.array([
    preprocess_image(
        row["image"]
    )
    for row in test_rows
])


# Labels
y_train = np.array([
    label_to_id[
        row["class_label"]
    ]
    for row in train_rows
])


y_test = np.array([
    label_to_id[
        row["class_label"]
    ]
    for row in test_rows
])


print(
    "\nTraining image shape:",
    X_train_img.shape
)

print(
    "Testing image shape:",
    X_test_img.shape
)


# ============================================================
# 9. DISPLAY SAMPLE IMAGES
# ============================================================

plt.figure(
    figsize=(12, 8)
)

for i in range(
    min(12, len(X_train_img))
):

    plt.subplot(
        3,
        4,
        i + 1
    )

    plt.imshow(
        X_train_img[i]
    )

    plt.title(
        CLASS_NAMES[
            y_train[i]
        ],
        fontsize=9
    )

    plt.axis("off")


plt.suptitle(
    "Sample Plant Leaf Images",
    fontsize=16
)

plt.tight_layout()
plt.show()


# ============================================================
# 10. CLASS DISTRIBUTION GRAPH
# ============================================================

train_counts = np.bincount(
    y_train,
    minlength=len(CLASS_NAMES)
)

plt.figure(
    figsize=(10, 5)
)

plt.bar(
    range(len(CLASS_NAMES)),
    train_counts
)

plt.xticks(
    range(len(CLASS_NAMES)),
    CLASS_NAMES,
    rotation=30,
    ha="right"
)

plt.xlabel(
    "Plant Disease Class"
)

plt.ylabel(
    "Number of Images"
)

plt.title(
    "Training Dataset Class Distribution"
)

plt.tight_layout()
plt.show()


# ============================================================
# PART A
# KNN USING IMAGE FEATURES
# ============================================================

print("\n" + "=" * 70)
print("               PART A: KNN")
print("=" * 70)


# ------------------------------------------------------------
# 11. IMAGE FEATURE EXTRACTION
# ------------------------------------------------------------

def extract_features(
    image
):

    # RGB mean
    mean_rgb = image.mean(
        axis=(0, 1)
    )

    # RGB standard deviation
    std_rgb = image.std(
        axis=(0, 1)
    )

    # Convert RGB to simple grayscale
    gray = image.mean(
        axis=2
    )

    mean_gray = gray.mean()

    std_gray = gray.std()

    # RGB color histograms
    histogram_features = []

    for channel in range(3):

        hist, _ = np.histogram(
            image[:, :, channel],
            bins=8,
            range=(0, 1)
        )

        hist = hist.astype(
            np.float32
        )

        hist = hist / (
            hist.sum() + 1e-8
        )

        histogram_features.extend(
            hist
        )

    features = np.concatenate([
        mean_rgb,
        std_rgb,
        [mean_gray, std_gray],
        np.array(
            histogram_features
        )
    ])

    return features


print(
    "\nExtracting image features..."
)

X_train_features = np.array([
    extract_features(image)
    for image in X_train_img
])

X_test_features = np.array([
    extract_features(image)
    for image in X_test_img
])

print(
    "Feature vector size:",
    X_train_features.shape[1]
)


# ------------------------------------------------------------
# 12. SCALE FEATURES
# ------------------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train_features
)

X_test_scaled = scaler.transform(
    X_test_features
)


# ------------------------------------------------------------
# 13. TRAIN KNN MODEL
# ------------------------------------------------------------

knn_model = KNeighborsClassifier(
    n_neighbors=5,
    weights="distance"
)

knn_model.fit(
    X_train_scaled,
    y_train
)

print(
    "KNN model trained successfully!"
)


# ------------------------------------------------------------
# 14. KNN PREDICTION
# ------------------------------------------------------------

knn_pred = knn_model.predict(
    X_test_scaled
)


# ------------------------------------------------------------
# 15. KNN PERFORMANCE
# ------------------------------------------------------------

knn_accuracy = accuracy_score(
    y_test,
    knn_pred
)

knn_precision = precision_score(
    y_test,
    knn_pred,
    average="weighted",
    zero_division=0
)

knn_recall = recall_score(
    y_test,
    knn_pred,
    average="weighted",
    zero_division=0
)

knn_f1 = f1_score(
    y_test,
    knn_pred,
    average="weighted",
    zero_division=0
)


print("\nKNN PERFORMANCE")

print(
    f"Accuracy  : {knn_accuracy:.4f}"
)

print(
    f"Precision : {knn_precision:.4f}"
)

print(
    f"Recall    : {knn_recall:.4f}"
)

print(
    f"F1 Score  : {knn_f1:.4f}"
)


print(
    "\nKNN Classification Report:"
)

print(
    classification_report(
        y_test,
        knn_pred,
        target_names=CLASS_NAMES,
        zero_division=0
    )
)


# ============================================================
# 16. KNN CONFUSION MATRIX
# ============================================================

knn_cm = confusion_matrix(
    y_test,
    knn_pred
)

fig, ax = plt.subplots(
    figsize=(9, 7)
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=knn_cm,
    display_labels=CLASS_NAMES
)

disp.plot(
    ax=ax,
    xticks_rotation=45,
    colorbar=False
)

plt.title(
    "KNN Confusion Matrix"
)

plt.tight_layout()
plt.show()


# ============================================================
# PART B
# CNN
# ============================================================

print("\n" + "=" * 70)
print("               PART B: CNN")
print("=" * 70)


# ------------------------------------------------------------
# 17. BUILD CNN MODEL
# ------------------------------------------------------------

cnn_model = tf.keras.Sequential([

    # Input
    tf.keras.layers.Input(
        shape=(64, 64, 3)
    ),

    # Convolution Block 1
    tf.keras.layers.Conv2D(
        32,
        (3, 3),
        activation="relu"
    ),

    tf.keras.layers.MaxPooling2D(
        (2, 2)
    ),

    # Convolution Block 2
    tf.keras.layers.Conv2D(
        64,
        (3, 3),
        activation="relu"
    ),

    tf.keras.layers.MaxPooling2D(
        (2, 2)
    ),

    # Convolution Block 3
    tf.keras.layers.Conv2D(
        128,
        (3, 3),
        activation="relu"
    ),

    tf.keras.layers.MaxPooling2D(
        (2, 2)
    ),

    # Fully connected section
    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(
        128,
        activation="relu"
    ),

    tf.keras.layers.Dropout(
        0.4
    ),

    # Four classes
    tf.keras.layers.Dense(
        len(CLASS_NAMES),
        activation="softmax"
    )
])


# ------------------------------------------------------------
# 18. COMPILE CNN
# ------------------------------------------------------------

cnn_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print(
    "\nCNN model created successfully!"
)

cnn_model.summary()


# ------------------------------------------------------------
# 19. TRAIN CNN
# ------------------------------------------------------------

print(
    "\nTraining CNN..."
)

history = cnn_model.fit(
    X_train_img,
    y_train,
    validation_split=0.20,
    epochs=10,
    batch_size=32,
    shuffle=True,
    verbose=1
)


# ------------------------------------------------------------
# 20. TEST CNN
# ------------------------------------------------------------

cnn_loss, cnn_accuracy = cnn_model.evaluate(
    X_test_img,
    y_test,
    verbose=0
)

print(
    f"\nCNN Test Accuracy: {cnn_accuracy:.4f}"
)


# ------------------------------------------------------------
# 21. CNN PREDICTIONS
# ------------------------------------------------------------

cnn_probabilities = cnn_model.predict(
    X_test_img,
    verbose=0
)

cnn_pred = np.argmax(
    cnn_probabilities,
    axis=1
)


# ------------------------------------------------------------
# 22. CNN PERFORMANCE
# ------------------------------------------------------------

cnn_precision = precision_score(
    y_test,
    cnn_pred,
    average="weighted",
    zero_division=0
)

cnn_recall = recall_score(
    y_test,
    cnn_pred,
    average="weighted",
    zero_division=0
)

cnn_f1 = f1_score(
    y_test,
    cnn_pred,
    average="weighted",
    zero_division=0
)


print(
    "\nCNN PERFORMANCE"
)

print(
    f"Accuracy  : {cnn_accuracy:.4f}"
)

print(
    f"Precision : {cnn_precision:.4f}"
)

print(
    f"Recall    : {cnn_recall:.4f}"
)

print(
    f"F1 Score  : {cnn_f1:.4f}"
)


print(
    "\nCNN Classification Report:"
)

print(
    classification_report(
        y_test,
        cnn_pred,
        target_names=CLASS_NAMES,
        zero_division=0
    )
)


# ============================================================
# 23. CNN CONFUSION MATRIX
# ============================================================

cnn_cm = confusion_matrix(
    y_test,
    cnn_pred
)

fig, ax = plt.subplots(
    figsize=(9, 7)
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cnn_cm,
    display_labels=CLASS_NAMES
)

disp.plot(
    ax=ax,
    xticks_rotation=45,
    colorbar=False
)

plt.title(
    "CNN Confusion Matrix"
)

plt.tight_layout()
plt.show()


# ============================================================
# 24. CNN TRAINING ACCURACY GRAPH
# ============================================================

plt.figure(
    figsize=(10, 5)
)

plt.plot(
    history.history["accuracy"],
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    marker="o",
    label="Validation Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.title(
    "CNN Training and Validation Accuracy"
)

plt.legend()

plt.grid(
    True,
    linestyle="--",
    alpha=0.6
)

plt.tight_layout()
plt.show()


# ============================================================
# 25. CNN LOSS GRAPH
# ============================================================

plt.figure(
    figsize=(10, 5)
)

plt.plot(
    history.history["loss"],
    marker="o",
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    marker="o",
    label="Validation Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.title(
    "CNN Training and Validation Loss"
)

plt.legend()

plt.grid(
    True,
    linestyle="--",
    alpha=0.6
)

plt.tight_layout()
plt.show()


# ============================================================
# PART C
# KNN VS CNN
# ============================================================

print("\n" + "=" * 70)
print("             KNN VS CNN COMPARISON")
print("=" * 70)


comparison = pd.DataFrame({

    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ],

    "KNN": [
        knn_accuracy,
        knn_precision,
        knn_recall,
        knn_f1
    ],

    "CNN": [
        cnn_accuracy,
        cnn_precision,
        cnn_recall,
        cnn_f1
    ]
})


print(
    comparison.round(4).to_string(
        index=False
    )
)


# ============================================================
# 26. COMPARISON GRAPH
# ============================================================

x = np.arange(
    len(comparison)
)

width = 0.35

plt.figure(
    figsize=(10, 6)
)

plt.bar(
    x - width / 2,
    comparison["KNN"],
    width,
    label="KNN"
)

plt.bar(
    x + width / 2,
    comparison["CNN"],
    width,
    label="CNN"
)

plt.xticks(
    x,
    comparison["Metric"]
)

plt.ylim(
    0,
    1.05
)

plt.xlabel(
    "Evaluation Metric"
)

plt.ylabel(
    "Score"
)

plt.title(
    "KNN vs CNN Performance Comparison"
)

plt.legend()

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.6
)

plt.tight_layout()
plt.show()


# ============================================================
# 27. SAMPLE PREDICTIONS
# ============================================================

sample_n = min(
    12,
    len(X_test_img)
)

fig, axes = plt.subplots(
    3,
    4,
    figsize=(14, 11)
)

axes = axes.ravel()


for i in range(
    sample_n
):

    axes[i].imshow(
        X_test_img[i]
    )

    actual = CLASS_NAMES[
        y_test[i]
    ]

    knn_result = CLASS_NAMES[
        knn_pred[i]
    ]

    cnn_result = CLASS_NAMES[
        cnn_pred[i]
    ]

    axes[i].set_title(
        "Actual: " + actual +
        "\nKNN: " + knn_result +
        "\nCNN: " + cnn_result,
        fontsize=8
    )

    axes[i].axis(
        "off"
    )


for i in range(
    sample_n,
    len(axes)
):

    axes[i].axis(
        "off"
    )


plt.suptitle(
    "Sample Plant Disease Predictions: KNN vs CNN",
    fontsize=16
)

plt.tight_layout()
plt.show()


# ============================================================
# 28. SAVE RESULTS
# ============================================================

OUTPUT_DIR = "/content/plant_project_outputs"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# Save comparison
comparison.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "knn_vs_cnn_results.csv"
    ),
    index=False
)


# Save CNN
cnn_model.save(
    os.path.join(
        OUTPUT_DIR,
        "plant_disease_cnn.keras"
    )
)


# Save KNN
import joblib

joblib.dump(
    knn_model,
    os.path.join(
        OUTPUT_DIR,
        "knn_model.pkl"
    )
)


# Save scaler
joblib.dump(
    scaler,
    os.path.join(
        OUTPUT_DIR,
        "feature_scaler.pkl"
    )
)


# Save metrics
metrics = pd.DataFrame({

    "Metric": [
        "KNN Accuracy",
        "KNN Precision",
        "KNN Recall",
        "KNN F1 Score",
        "CNN Accuracy",
        "CNN Precision",
        "CNN Recall",
        "CNN F1 Score"
    ],

    "Value": [
        knn_accuracy,
        knn_precision,
        knn_recall,
        knn_f1,
        cnn_accuracy,
        cnn_precision,
        cnn_recall,
        cnn_f1
    ]
})

metrics.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "model_metrics.csv"
    ),
    index=False
)


# ============================================================
# FINAL MESSAGE
# ============================================================

print("\n" + "=" * 70)
print("           PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    "\nOutput files saved in:"
)

print(
    OUTPUT_DIR
)

print(
    "\nUse the ACTUAL KNN and CNN results from this run"
)

print(
    "in your project report."
)
```
